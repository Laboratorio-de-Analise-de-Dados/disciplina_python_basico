## 01 – Obtenha os resultados e retorne o código Python dos seguintes cálculos:

<br>a = 2.0 + 3.0 / 5.0 == 10
<br>b = (a + 10.2) - 5.0 * 2.0 <= 2.9
<br>c = 10 <= a / b + 33.8 < 100

```python
a = 2.0 + 3.0 / 5.0 == 10
b = (a + 10.2) - 5.0 * 2.0 <= 2.9
c = 10 <= a / b + 33.8 < 100
d = 5 != a / (b + c)
e = 33 == d * 2

print(f"a: {a}")
print(f"b: {b}")
print(f"c: {c}")
print(f"d: {d}")
print(f"e: {e}")
```

## 02 – Altere os operadores relacionais para modificar o resultado de cada uma das comparações.

<br>('a',1,True,'b',0,False) != ('a',True,1,'b',False,0)
<br>['a',1,True,'b',0,False] != ('a',True,1,'b',False,0)
<br>['a',1,True,'b',0,False] != ['a',True,1,'b',False,0]
<br>['a',1,True,'b',0,False] > ['a',True,1]
<br>['a',1,True,'b',0,False] == 1
<br>['a',True,1] <= ['a',True,1] < ['a',1,True,'b',0,False]

```python
bool_1 = ('a',1,True,'b',0,False) != ('a',True,1,'b',False,0)
bool_2 = ['a',1,True,'b',0,False] != ('a',True,1,'b',False,0)
bool_3 = ['a',1,True,'b',0,False] != ['a',True,1,'b',False,0]
bool_4 = ['a',1,True,'b',0,False] > ['a',True,1]
bool_5 = ['a',1,True,'b',0,False] == 1
bool_6 = ['a',True,1] <= ['a',True,1] < ['a',1,True,'b',0,False]

print(f"bool_1: {bool_1}")
print(f"bool_2: {bool_2}")
print(f"bool_3: {bool_3}")
print(f"bool_4: {bool_4}")
print(f"bool_5: {bool_5}")
print(f"bool_6: {bool_6}")
```

```python
bool_1 = ('a',1,True,'b',0,False) == ('a',True,1,'b',False,0)
bool_2 = ['a',1,True,'b',0,False] == ('a',True,1,'b',False,0)
bool_3 = ['a',1,True,'b',0,False] == ['a',True,1,'b',False,0]
bool_4 = ['a',1,True,'b',0,False] < ['a',True,1]
bool_5 = ['a',1,True,'b',0,False] != 1
bool_6 = ['a',True,1] > ['a',True,1] >= ['a',1,True,'b',0,False]

print(f"bool_1: {bool_1}")
print(f"bool_2: {bool_2}")
print(f"bool_3: {bool_3}")
print(f"bool_4: {bool_4}")
print(f"bool_5: {bool_5}")
print(f"bool_6: {bool_6}")
```
