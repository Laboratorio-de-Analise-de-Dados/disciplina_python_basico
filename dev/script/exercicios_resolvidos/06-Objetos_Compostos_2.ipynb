## 10 – Importe o data.csv (dados de seguro na Flórida em 2012) para uma DataFrame. Armazene em um dicionário o número de entradas, uma lista com as classes das colunas e outra lista com o número de elementos de cada classe, três listas, contendo a média, o desvio padrão e o terceiro quartil da primeira, quarta e nona colunas . Recupere o dicionário.

```python
df = pd.read_csv('./data/data.csv')
df.head()
```

```python
dict_dados = {'num_entradas': len(df)}
dict_dados
```

```python
df.dtypes
```

```python
df.dtypes.value_counts()
```

```python
dict_dados['classes_colunas'] = df.dtypes.value_counts().index.astype(str).tolist()
dict_dados['num_elementos_por_classe'] = df.dtypes.value_counts().values.astype(str).tolist()
dict_dados
```

```python
colunas = df.columns[[0,3,8]]
df[colunas].describe()
```

```python
medidas_resumo = df[colunas].describe().iloc[[1,2,6]]
medidas_resumo
```

```python
dict_dados['media'] = medidas_resumo.loc['mean', :].to_list()
dict_dados['desvio'] = medidas_resumo.loc['std', :].to_list()
dict_dados['quartio'] = medidas_resumo.loc['75%', :].to_list()

dict_dados
```
