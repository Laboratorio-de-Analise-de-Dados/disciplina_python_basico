## Instancie um objeto da classe lista contendo 5.000 valores inteiros iniciando em 0. 
Quantos são pares e quantos são impares?

## 01 – Descreva o racional da sua resposta:

R.: Posso utilizar uma estrutura de repetição que é finalizada quando meu objeto tiver 5000 números. Com uma segunda estrutura, itero os itens no objeto. Divido cada um por 2 e obtenho o resto da divisão com o operador aritmético %. Quando o resultado for 0, isso significa que o número é par, quando for 1, é impar. Uso uma estrutura condicional para cada um dos valores e incremento um objeto inteiro para os pares e outro para os impares. Recupero com print.

## 03 – Escreva o programa em Python:

```python
lista_num = []
for conta in range(5_000):
    lista_num.append(conta)
len(lista_num)
```

```python
soma_par = 0
soma_impar = 0
for num in lista_num:
    if num % 2 == 0:
        soma_par += 1
    else:
        soma_impar += 1
print(f'Soma dos números pares: {soma_par:,.2f}'.replace(',', "_").replace('.', ",").replace('_', '.'))
print(f'Soma dos números ímpares: {soma_impar:,.2f}'.replace(',', "_").replace('.', ",").replace('_', '.'))
```

## 04 – Substitua a estrutura de repetição e responda novamente a pergunta anterior.

```python
soma_par = 0
soma_impar = 0
conta = 0
while conta < len(lista_num):
    if lista_num[conta] % 2 == 0:
        soma_par += 1
    else:
        soma_impar += 1
    conta += 1
print(f'Soma dos números pares: {soma_par:,.2f}'.replace(',', "_").replace('.', ",").replace('_', '.'))
print(f'Soma dos números ímpares: {soma_impar:,.2f}'.replace(',', "_").replace('.', ",").replace('_', '.'))
```

```python
import numpy as np
```

```python
# Resposta alternativa sem o uso de loops
array_num = np.array(lista_num)
soma_par_array = len(array_num[array_num % 2 == 0])
soma_impar_array = len(array_num[array_num % 2 != 0])

print(f'Soma dos números pares (array): {soma_par_array:,.2f}'.replace(',', "_").replace('.', ",").replace('_', '.'))
print(f'Soma dos números ímpares (array): {soma_impar_array:,.2f}'.replace(',', "_").replace('.', ",").replace('_', '.'))
```

## 08 – Substitua a estrutura de repetição e responda novamente a pergunta anterior.

```python
lista_par_impar_genes = [str(item) + " é par" if item % 2 == 0 else str(item) + " é impar" for item in df_genes['num_letras']]

conta = 0
while conta < len(lista_par_impar_genes):
    print(lista_par_impar_genes[conta])
    conta += 1
```
