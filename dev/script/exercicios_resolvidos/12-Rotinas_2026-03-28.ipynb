## 04 – Escreva o programa em Python:

```python
import numpy as np
```

```python
# Instânciando o objeto array do numpy
array_randomico = np.random.normal(loc=10, scale=3, size=5)
array_randomico
```

```python
# Número de valores são superiores a 10
superior_a_10 = 0
alvo = 10
if array_randomico[0] > alvo:
    print(f"O valor {array_randomico[0]} é superior a {alvo}")
    superior_a_10 += 1
if array_randomico[1] > alvo:
    print(f"O valor {array_randomico[1]} é superior a {alvo}")
    superior_a_10 += 1
if array_randomico[2] > alvo:
    print(f"O valor {array_randomico[2]} é superior a {alvo}")
    superior_a_10 += 1
if array_randomico[3] > alvo:
    print(f"O valor {array_randomico[3]} é superior a {alvo}")
    superior_a_10 += 1
if array_randomico[4] > alvo:
    print(f"O valor {array_randomico[4]} é superior a {alvo}")
    superior_a_10 += 1
print(f"O número de valores superiores a {alvo} é: {superior_a_10}")
```

```python
def verificar_superior_a_10(array: np.array, alvo: int = 10) -> None:
    ''' 
    Verifica quais elementos do array são
    superiores a um valor alvo e conta quantos são.
        Parâmetros:
            array (np.array): O array a ser verificado.
            alvo (int): O valor de referência para comparação (padrão é 10).
        Retorna:
            None: A função apenas imprime os resultados,
                    não retorna nenhum valor.
    '''
    superior_a_10 = 0
    for elemento in array:
        if elemento > alvo:
            print(f"O valor {elemento} é superior a {alvo}")
            superior_a_10 += 1
    print(f"O número de valores superiores a {alvo} é: {superior_a_10}")
    return None
```

```python
help(verificar_superior_a_10)
```

```python
verificar_superior_a_10(array=array_randomico)
```

## Desenvolva um programa de computador que retorne a média de um conjunto de valores.

## 06 – Descreva o racional da sua resposta:

R.: Declaração de objetos com valores. A função receberá estes valores, usando argumentos como parâmetro, o que possibilita receber qualquer número de objetos. Na função, primeiro declaramos um objeto inteiro que receberá a soma dos itens recebidos. Estes serão então iterados, incrementando o objeto que soma os valores recebidos. O retorno da função será o valor da soma dividido pelo número de elementos.

## 07 – Escreva o programa em Python:

```python
def calcular_media_valores(*valores: int) -> float:
    '''
        Calcula a média de um conjunto de valores inteiros.
            Parâmetros:
                *valores (int): Um número variável de argumentos inteiros.
            Retorna:
                float: A média dos valores fornecidos.
    '''
    soma_valores = 0
    n_valores = 0
    for valor in valores:
        soma_valores += int(valor)
        n_valores += 1
    media = soma_valores / n_valores
    return media
```

## 09 – Desenvolva um programa de computador que retorne a divisão de dois objetos simples. Seu programa deve ser capaz de reconhecer a classe destes objetos e o valor, antes de retornar ao usuário.

```python
def verificar_tipos_simples(objeto: int | float) -> bool:
    '''
        Verifica se os objetos fornecidos são do tipo numpy.ndarray,
        pandas.Series e pandas.DataFrame.
            Parâmetros:
                objetos (int | float): Um objeto simples a serem verificado.
            Retorna:
                bool: True se todos os objetos forem do tipo esperado, 
                    False caso contrário.
    '''
    tipo_int = type(objeto) == int
    tipo_float = type(objeto) == float
    
    if tipo_int or tipo_float:
        print(f"O objeto {objeto} é do tipo esperado.")
        return True
    else:
        print(f"O objeto {objeto} não é do tipo esperado.")
        return False
```

```python
help(verificar_tipos_simples)
```

```python
array_1 = np.random.normal(loc=10, scale=3, size=(5, 3))
array_2 = np.random.normal(loc=10, scale=3, size=(5, 3))
type(array_2)
```

```python
import pandas as pd
```

```python
serie_1 = pd.Series(array_1[0])
serie_2 = pd.Series(array_2[1])
type(serie_1)
```

```python
df_1 = pd.DataFrame(data=array_1, columns=['A', 'B', 'C'])
df_2 = pd.DataFrame(data=array_2, columns=['A', 'B', 'C'])
type(df_1)
```

```python
objetos = {
    "array_1": array_1,
    "array_2": array_2,
    "serie_1": serie_1,
    "serie_2": serie_2,
    "df_1": df_1,
    "df_2": df_2,
    "string": f"{serie_1.iloc[0]}",
    "inteiro": int(serie_1.iloc[0]),
    "float": array_1[0, 0],
    "lista": list(array_1[0]),
    "tuple": tuple(array_1[0]),
    "set": set(array_1[0]),
}
objetos
```

```python
for tipo, objeto in enumerate(objetos.items()):
    print("-"*10)
    print(f"{objeto[0]} - {verificar_tipos_simples(objeto[1])}")
```

```python
def dividir_valores_simples(num1: int | float, num2: int | float) -> float:
    '''
        Realiza a divisão de dois números, tratando a divisão por zero.
            Parâmetros:
                num1 (int | float): O numerador da divisão.
                num2 (int | float): O denominador da divisão.
            Retorna:
                float: O resultado da divisão se num2 for diferente de zero,
                    ou uma mensagem de erro caso contrário.
    '''
    if num2 == 0:
        print("Erro: Divisão por zero não é permitida.")
        return None
    else:
        if verificar_tipos_simples(num1) and verificar_tipos_simples(num2):
            resultado = num1 / num2
            print(f"O resultado da divisão de {num1} por {num2} é: {resultado}")
            return resultado
        else:
            print("Erro: Ambos os números devem ser do tipo int ou float.")
            return None
```

```python
help(dividir_valores_simples)
```

```python
teste_divisao = {
    "int/int": (40, 2),
    "float/float": (40.0, 2.0),
    "int/float": (40, 2.0),
    "float/int": (40.0, 2),
    "erro_1": (40, 0),
    "erro_2": ([40], 2),
    "erro_3": ([40], [2]),
    "erro_4": (40, [2]),
}
teste_divisao
```

```python
for k, v in teste_divisao.items():
    print("-"*10)
    print(f"{k}")
    dividir_valores_simples(v[0], v[1])
```

## 10 – Desenvolva um programa de computador que retorne a divisão de dois objetos compostos. Seu programa deve ser capaz de reconhecer a classe destes objetos e o valor, antes de retornar ao usuário.

```python
def verificar_tipos_compostos(
        objeto: np.ndarray | pd.Series | pd.DataFrame) -> bool:
    '''
        Verifica se os objetos fornecidos são do tipo numpy.ndarray,
        pandas.Series e pandas.DataFrame.
            Parâmetros:
                objetos (list): Uma lista de objetos a serem verificados.
            Retorna:
                bool: True se todos os objetos forem do tipo esperado, 
                    False caso contrário.
    '''
    tipo_array = type(objeto) == np.ndarray
    tipo_serie = type(objeto) == pd.Series
    tipo_frame = type(objeto) == pd.DataFrame
    
    if tipo_array or tipo_serie or tipo_frame:
        print("O objeto é do tipo esperado.")
        return True
    else:
        print("O objeto não é do tipo esperado.")
        return False
```

```python
help(verificar_tipos_compostos)
```

```python
for tipo, objeto in enumerate(objetos.items()):
    print("-"*10)
    print(f"{objeto[0]} - {verificar_tipos_compostos(objeto[1])}")
```

```python
def dividir_valores_compostos(
        comp1: np.ndarray | pd.Series | pd.DataFrame,
        comp2: np.ndarray | pd.Series | pd.DataFrame) -> float:
    '''
        Realiza a divisão de dois números, tratando a divisão por zero.
            Parâmetros:
                comp1 (np.ndarray | pd.Series | pd.DataFrame): O numerador da divisão.
                comp2 (np.ndarray | pd.Series | pd.DataFrame): O denominador da divisão.
            Retorna:
                float: O resultado da divisão se comp2 for diferente de zero,
                    ou uma mensagem de erro caso contrário.
    '''
    if verificar_tipos_compostos(comp1) and verificar_tipos_compostos(comp2):
        resultado = comp1 / comp2
        print(f"O resultado da divisão de {comp1} por {comp2} é: {resultado}")
        return resultado
    else:
        print("Erro: Ambos os números devem ser do tipo array, Series ou DataFrame.")
        return None
```

```python
for k, v in teste_divisao.items():
    print("-"*10)
    print(f"{k}")
    dividir_valores_compostos(v[0], v[1])
```
