## 04 – Escreva o programa em Python:

```python
import numpy as np
```

```python
# Instânciando o objeto array do numpy
array_randomico = np.random.normal(loc=10, scale=3, size=5)
array_randomico
```

```python
# Número de valores são superiores a 10
superior_a_10 = 0
alvo = 10
if array_randomico[0] > alvo:
    print(f"O valor {array_randomico[0]} é superior a {alvo}")
    superior_a_10 += 1
if array_randomico[1] > alvo:
    print(f"O valor {array_randomico[1]} é superior a {alvo}")
    superior_a_10 += 1
if array_randomico[2] > alvo:
    print(f"O valor {array_randomico[2]} é superior a {alvo}")
    superior_a_10 += 1
if array_randomico[3] > alvo:
    print(f"O valor {array_randomico[3]} é superior a {alvo}")
    superior_a_10 += 1
if array_randomico[4] > alvo:
    print(f"O valor {array_randomico[4]} é superior a {alvo}")
    superior_a_10 += 1
print(f"O número de valores superiores a {alvo} é: {superior_a_10}")
```
