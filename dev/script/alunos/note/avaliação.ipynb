```python
def get_alunos(aluno: str) -> str:
    try:
        aluno_nome = aluno.split("https")[0]
        aluno_nome = aluno_nome[:-2]
    except:
        aluno_nome = aluno
    return aluno_nome
```

```python
df_exercicios['alunos'] = [get_alunos(i) for i in df_exercicios['Alunos']]
df_exercicios.head()
```

```python
df_exercicios.drop(columns=['Alunos', 'Status', 'Revisor', 'Total de questões', 'Acertos'], inplace=True)
df_exercicios.rename(columns={'Name': 'exercicio'}, inplace=True)
df_exercicios.head()
```

```python
with open("./alunos/data/Email_Alunos.txt", "r") as f:
    data_email = f.readlines()
data_email = data_email[1:]
data_email
```

```python
with open("./alunos/data/GitHub_Alunos.txt", "r") as f:
    data_git = f.readlines()
data_git = data_git[1:]
data_git
```

```python
lista_alunos_email = [v[3:].strip().split(" - ") if k < 10 else v[4:].strip().split(" - ") for k, v in enumerate(data_email)]
lista_alunos_email[:4]
```

```python
lista_alunos_git = [v[3:].strip().split(" - ") if k < 10 else v[4:].strip().split(" - ") for k, v in enumerate(data_git)]
lista_alunos_git[:4]
```

```python
data = {
    "alunos": [i[0] for i in lista_alunos_git],
    "github": [i[1] for i in lista_alunos_git],
}
df = pd.DataFrame(data)
df.head()
```

```python
df_aulas = df.copy()
for i in exercicios:
    df_aulas.loc[:, i] = False
df_aulas.head()
```

```python
c = 1
url = df_aulas['github'][c]
print(url)
response = requests.get(url)
if response.status_code == 200:
    data = response.text.split("name")
data = [i for i in data if '.ipynb","' in i]
data = [i.split('","path')[0][3:] for i in data]
data
```
