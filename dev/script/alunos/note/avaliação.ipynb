# Funções

```python
def get_alunos(aluno: str) -> str:
    try:
        aluno_nome = aluno.split("https")[0]
        aluno_nome = aluno_nome[:-2]
    except:
        aluno_nome = aluno
    return aluno_nome
```

# Importando banco

```python
# Lendo o arquivo CSV
df_exercicios = pd.read_csv("./alunos/data/relatorio_final_2026-05-02.csv")

# Extraindo os nomes dos alunos e renomeando as colunas
df_exercicios['alunos'] = [get_alunos(i) for i in df_exercicios['Alunos']]
df_exercicios.rename(columns={'Name': 'exercicio'}, inplace=True)

# Removendo colunas desnecessárias e reorganizando o DataFrame
df_exercicios.drop(columns=[
    "Acertos",
    "Total de questões",
    "Revisor",
    "Status",
    "Alunos",
], inplace=True)
df_exercicios.rename(columns={'Nota Final': 'notas'}, inplace=True)
df_exercicios = df_exercicios[["alunos", "exercicio", "notas"]]
df_exercicios.dropna(inplace=True)
df_exercicios.head()
```

```python
df_exercicios.info()
```

```python
# Editando a entrada "02 - Introducao " para "02 - Introducao"
index = df_exercicios[df_exercicios['exercicio'] == '02 - Introducao '].index
df_exercicios.loc[index, 'exercicio'] = '02 - Introducao'

# Editando a entrada "14-  Variaveis Quantitativas" para "14 - Variaveis Quantitativas"
index = df_exercicios[df_exercicios['exercicio'] == '14-  Variaveis Quantitativas'].index
df_exercicios.loc[index, 'exercicio'] = '14 - Variaveis Quantitativas'

# Editando a entrada "Natalia Zureck Xaviern" para "Natalia Zureck Xavier"
index = df_exercicios[df_exercicios['alunos'] == 'Natalia Zureck Xaviern'].index
df_exercicios.loc[index, 'alunos'] = 'Natalia Zureck Xavier'

# Editando a entrada "87,5" para 87.5
index = df_exercicios[df_exercicios['notas'] == "87,5"].index
df_exercicios.loc[index, 'notas'] = "87.5"

# Editando a entrada "190,0" para 90.0
index = df_exercicios[df_exercicios['notas'] == "190"].index
df_exercicios.loc[index, 'notas'] = 90

# Obtendo uma lista com os nomes dos exercícios 
exercicios = df_exercicios['exercicio'].value_counts().sort_values().index

# Obtendo uma lista com os nomes dos exercícios ordenados por número
exer_lista = []
for j in range(1, len(exercicios) + 5):
    for i in exercicios:
        if f"0{j}" == f"{i}".split(" - ")[0]:
            exer_lista.append(i)
        elif f"{j}" == f"{i}".split(" - ")[0]:
            exer_lista.append(i)
ut.printLis(exer_lista)
```

```python
# Verificando as colunas da DataFrame
df_exercicios.head()
```

```python
# Modificando o tipo da coluna "nota" para float
df_exercicios['notas'] = df_exercicios['notas'].astype(float)
df_exercicios.info()
```

```python
df_notas = (
    df_exercicios
    .groupby("alunos")
    .notas
    .mean()
    .round(2)
    .to_frame()
    .reset_index()
)
df_notas
```

```python
df_exercicios['num_exercicios'] = df_exercicios['notas'] == 0.0
df_exercicios.head()
```

```python
# Instanciando uma DataFrame com os nomes dos alunos e o número de exercícios realizados
df_num_exercicios = (
    df_exercicios
    .groupby("alunos")
    .num_exercicios
    .sum()
    .to_frame()
    .reset_index()
    # .assign(num_exercicios=lambda x: x-14)
)
df_num_exercicios['num_exercicios'] = df_num_exercicios['num_exercicios'] - 14
df_num_exercicios['num_exercicios'] = df_num_exercicios['num_exercicios'] * -1
df_num_exercicios
```

# Buscando e-mail dos alunos

```python
# Carrega o arquivo de emails dos alunos
with open("./alunos/data/Email_Alunos.txt", "r") as f:
    data_email = f.readlines()
data_email
```

```python
# Instancando uma lista com os nomes dos alunos a partir do arquivo de emails
lista_alunos_email = [v[3:].strip().split(" - ") if k < 10 else v[4:].strip().split(" - ") for k, v in enumerate(data_email)]
lista_alunos_nomes = [i[0] for i in lista_alunos_email]
ut.printLis(lista_alunos_nomes)
```

```python
# Instanciando uma lista com os emails dos alunos a partir do arquivo de emails
lista_alunos_email = [i[1] for i in lista_alunos_email]
ut.printLis(lista_alunos_email)
```

# Buscando os url dos alunos

```python
# Instanciaando uma lista com os nomes dos alunos e os url dos GitHub
with open("./alunos/data/GitHub_Alunos.txt", "r") as f:
    data_git = f.readlines()
data_git = data_git[1:]
data_git
```

```python
# Instancando uma lista com os nomes e os url do GitHub
lista_alunos_git = [v[3:].strip().split(" - ") if k < 10 else v[4:].strip().split(" - ") for k, v in enumerate(data_git)]
lista_alunos_git[:4]
```

# Agrupando os dados

```python
# Instanciando uma DataFrame com os nomes dos alunos e os url do GitHub
data = {
    "alunos": [i[0] for i in lista_alunos_git],
    "github": [i[1] for i in lista_alunos_git],
}
df_git = pd.DataFrame(data)
df_git.head()
```

```python
# Instanciando uma DataFrame com os nomes dos alunos e os emails
data = {
    "alunos": lista_alunos_nomes,
    "email": lista_alunos_email,
}
df_mail = pd.DataFrame(data)
df_mail.head()
```

```python
df_identificador = df_git.merge(df_mail, on="alunos", how="left")
df_exercicios_nomes = df_identificador.merge(df_num_exercicios, on="alunos", how="left")
df_alunos = df_exercicios_nomes.merge(df_notas, on="alunos", how="left")

df_alunos
```
