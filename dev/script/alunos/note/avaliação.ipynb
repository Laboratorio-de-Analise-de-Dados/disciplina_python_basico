```python
def get_alunos(aluno: str) -> str:
    try:
        aluno_nome = aluno.split("https")[0]
        aluno_nome = aluno_nome[:-2]
    except:
        aluno_nome = aluno
    return aluno_nome
```

```python
def atribuir_conceito(nota: float) -> str:
    '''
        Conceito A - Entre 90% e 100% (rendimento excelente);
        Conceito B - Entre 75% e 89% (rendimento bom);
        Conceito C - Entre 60% e 74% (rendimento regular);
        Conceito D - Inferior a 60% (rendimento insuficiente).
    '''
    if 90 <= nota <= 100:
        return "A"
    elif 75 <= nota <= 89:
        return "B"
    elif 60 <= nota <= 74:
        return "C"
    elif nota < 60:
        return "D"
```

```python
def atualizar_notas(
            aluno: int,
            notas_final: list[int]
    ) -> None:
    '''
        Função para atualizar as notas dos exercícios de um aluno específico.
                Args:
                        aluno (int): Índice do aluno na DataFrame df_alunos.
                        notas_final (list[int]): Lista com as notas finais dos
                        exercícios do aluno.
                Retorna:
                        None
    '''
#     Buscando index dos exercicios do aluno específico
    index = (
        df_exercicios[df_exercicios['alunos'] == df_alunos['alunos'][aluno]]
        .sort_values(by="exercicio")
        .index
    )

#     Atualizando as notas dos exercícios do aluno específico
    df_exercicios.loc[index, 'notas'] = notas_final

#     Instanciando lista de index das colunas de exercícios
    colunas = [7,8,9,10,11,12,13][::-1]
    colunas += [14,15,16,17,18,19,20][::-1]

#     Atualizando as notas dos exercícios do aluno específico na df_alunos
    df_alunos.iloc[aluno, 7:] = df_exercicios.loc[index, 'notas'].values

#     Visualizando as notas dos exercícios do aluno específico
    display(df_alunos[df_alunos['alunos'] == df_alunos['alunos'][aluno]])
    return None
```
