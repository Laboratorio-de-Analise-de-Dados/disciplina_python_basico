```python
import os
import pandas as pd
import numpy as np
import requests
```

```python
os.listdir()
```

```python
exercicios = [i[:-6] for i in os.listdir("exercicios_resolvidos") if i.endswith(".ipynb")]
exercicios
```

```python
with open("./alunos/GitHub_Alunos.txt", "r") as f:
    data = f.readlines()
data = data[1:]
data
```

```python
lista_alunos = [v[3:].strip().split(" - ") if k < 10 else v[4:].strip().split(" - ") for k, v in enumerate(data)]
lista_alunos[:4]
```

```python
data = {
    "nomes": [i[0] for i in lista_alunos],
    "github": [i[1] for i in lista_alunos],
}
df = pd.DataFrame(data)
df.head()
```

```python
df_aulas = df.copy()
for i in exercicios:
    df_aulas.loc[:, i] = False
df_aulas.head()
```

```python
df_aulas.info()
```

```python
url = df_aulas['github'][0]
print(url)
response = requests.get(url)
if response.status_code == 200:
    data = response.text.split("name")
data = [i for i in data if '.ipynb","' in i]
data = [i.split('","path')[0][3:] for i in data]
data
```
