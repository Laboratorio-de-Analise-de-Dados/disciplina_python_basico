```python
# Instanciar um objeto complexo
obj_complex_1 = complex(3, 4)  # Creates (3 + 4j)
obj_complex_2 = complex(5)    # Creates (5 + 0j)
obj_complex_3 = complex("2+3j") # Creates (2 + 3j)

print(obj_complex_1)  # Output: (3+4j)
print(obj_complex_2)  # Output: (5+0j)
print(obj_complex_3)  # Output: (2+3j)
```

```python
# Instanciando bytes
ascii_code = [72, 101, 108, 108, 111] # ASCII code
obj_bytes = bytes(ascii_code)

print(ascii_code)
print(obj_bytes)
print(obj_bytes)
print(obj_bytes.decode('utf8'))
```
