```python
def highlight_max(s):
    is_max = s == s.max()
    return ['background-color: green' if v else '' for v in is_max]
```

```python
(
    df[df['Sample'] == 'Sample 1']
    .style
    .apply(highlight_max, axis=0)
)
```
