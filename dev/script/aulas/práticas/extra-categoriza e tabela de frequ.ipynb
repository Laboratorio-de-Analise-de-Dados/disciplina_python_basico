```python
df = pd.DataFrame({'variavel': np.random.normal(loc=0, scale=1, size=100)})
df.head()
```

```python
df.describe()
```

```python
bins = [-3, -1.2, -1, 0, 1.2, 2, 3]
df['ordinal'] = pd.cut(df['variavel'], bins=bins, labels=['Cat A', 'Cat B', 'Cat C', 'Cat D', 'Cat E', 'Cat F'])
df.head()
```

```python
df['nominal'] = pd.qcut(df['variavel'], q=6, labels=['Q1', 'Q2', 'Q3', 'Q4', 'Q5', 'Q6'])
df.head()
```

```python
df.head()
```

```python
freq_nominal = (
    df['nominal'].value_counts()
    .sort_index()
    .to_frame()
    .rename(columns={'nominal': 'abso_simp'})
    .assign(relat_simp=df['nominal'].value_counts(normalize=True))
)
freq_nominal
```

```python
freq_ordinal = (
    df['ordinal'].value_counts()
    .sort_index()
    .to_frame()
    .rename(columns={'ordinal': 'abso_simp'})
    .assign(relat_simp=df['ordinal'].value_counts(normalize=True))
)
freq_ordinal
```

```python
freq_ordinal = (
    freq_ordinal.assign(absol_acum=freq_ordinal['abso_simp'].cumsum())
    .assign(relat_acum=freq_ordinal['relat_simp'].cumsum())
)
freq_ordinal
```

```python
def gerar_cepa(codigo: str) -> list:
    num = np.random.randint(300, 500)
    cepa = f"{codigo}_"*num
    cepa = cepa.split("_")[:-1]
    return cepa
```

```python
cepas = [
    "Cepa A",
    "Cepa B",
    "Cepa C",
    "Cepa D",
    "Cepa E",
    "Cepa F",
    "Total",
]
cepa_final = [gerar_cepa(cepa) for cepa in cepas]
cepa = np.array(
                cepa_final[0] +
                cepa_final[1] +
                cepa_final[2] +
                cepa_final[3] +
                cepa_final[4] +
                cepa_final[5]
)
cepa.shape
```

```python
cepa[:10]
```

```python
np.random.shuffle(cepa)
cepa[:10]
```

```python
df = pd.DataFrame(cepa, columns=["Cepa"])
df.head()
```

```python
tabela_freq = (
                df.value_counts()
                .sort_index()
                .to_frame()
                .rename(columns={0: "absol_simp"})
                .assign(
                    relat_simp=df['Cepa'].value_counts(normalize=True).values
                )
                .round(3)
)
soma = tabela_freq.sum().to_frame().T
tabela_nominal = pd.concat((tabela_freq, soma), axis=0)
tabela_nominal.index = cepas
tabela_nominal
```

```python
tabela_ordinal = (
                tabela_nominal.assign(
                    absol_acum=tabela_nominal["absol_simp"].cumsum(),
                    relat_acum=tabela_nominal["relat_simp"].cumsum()
                )
)
tabela_ordinal.loc['Total', ['absol_acum', 'relat_acum']] = np.nan
tabela_ordinal
```
