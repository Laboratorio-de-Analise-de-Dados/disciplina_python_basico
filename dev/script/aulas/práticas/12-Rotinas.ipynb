```python
def soma(num1, num2):
    ''' Função que recebe dois valores e retorna sua soma'''
    if type(num1) != str and type(num2) != str:
        return num1+num2
    else:
        return "Valores não são inteiros"
```

```python
help(soma)
```

```python
print(soma(20, "5.1"))
```

```python
for _ in range(200):
    print(soma(60, 10))
```

```python
def fazer(mensagem: str) -> str:
    '''
        Função que recebe uma mensagem e retorna a mesma mensagem em caixa alta
            Args:
                mensagem (str): A mensagem a ser convertida para caixa alta
            Returns:
                str: A mensagem convertida para caixa alta
    '''
    retorno = mensagem.upper()
    return retorno
```

```python
print(fazer(mensagem="Python lindo"))
```
